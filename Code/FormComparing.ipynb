````python
#распределение токов
def plot_current_distribution(currents, freq_index, layer_index, frequencies, Params, title_prefix=""):
    """
    Function to display a current distribution in a given layer of the structure for a given frequency.

    Parameters:
    currents: array of currents for each frequency.
    freq_index: index of the frequency to display.
    layer_index: index of the layer to display (from 1 to Nz).
    frequencies: array of frequencies (in MHz).
    Params: dictionary with model parameters, must contain 'shape' and 'Orientations'.
    title_prefix: prefix for the plot title (default is an empty string).
    """

    current_at_freq = currents[freq_index]

    shape_string = Params['Shape']
    shape_list = shape_string.split('x')
    orientations = Params.get('Orientations', 'zyx')
    packing_type = Params['Packing']

    shape_dict = dict(zip(orientations, [int(dim) for dim in shape_list]))
    Nx = shape_dict['x']
    Ny = shape_dict['y']
    Nz = shape_dict['z']

    if layer_index < 0 or layer_index > Nz:
        print("Индекс слоя вне диапазона.")
        return

    layer_size = Nx * Ny

    start_index = layer_index * layer_size
    end_index = (layer_index + 1) * layer_size
    layer_currents_flat = current_at_freq[start_index:end_index]

    layer_currents_matrix = layer_currents_flat.reshape((Ny, Nx))

    if packing_type == 'Ellipse':
        # Calculate ellipse parameters
        dz, dy, dx = Params['Dz'], Params['Dy'], Params['Dx']
        Nz_total, Ny_total, Nx_total = Params['N']['z']['nz'], Params['N']['y']['ny'], Params['N']['x']['nx']
        
        R_z, R_y, R_x = (Nz_total-1)*dz/2, (Ny_total-1)*dy/2, (Nx_total-1)*dx/2
        
        # Adjust radii for different packing cases
        if len(orientations) == 1:
            pass
        elif Params['N']['z']['nz'] == Params['N']['y']['nz']:  # center system
            R_z += dz/2
            R_y += dy/2
            R_x += dx/2
        elif Params['N']['z']['nz'] < Params['N']['y']['nz']:  # open system
            R_z += dz 
            R_y += dy
            R_x += dx
        
        # Create mask for cells inside the ellipse
        mask = np.ones((Ny, Nx), dtype=bool)
        for j in range(Ny):
            for i in range(Nx):
                # Calculate global coordinates (need to adjust based on your coordinate system)
                # This depends on how the ellipse is positioned in your system
                x_pos = i * dx + dx/2  # assuming center of cell
                y_pos = j * dy + dy/2
                z_pos = layer_index * dz + dz/2  # for current layer
                
                distance = ((x_pos-R_x)/R_x)**2 + ((y_pos-R_y)/R_y)**2 + ((z_pos-R_z)/R_z)**2
                if distance > 1.0:
                    mask[j, i] = False
        
        # Apply mask to currents
        layer_currents_matrix = np.where(mask, layer_currents_matrix, np.nan)

    current_magnitude = np.abs(layer_currents_matrix)

    extent = [0.5, Nx + 0.5, 0.5, Ny + 0.5]

    plt.imshow(current_magnitude, cmap='viridis', extent=extent, origin='lower')

    if packing_type == 'Ellipse':
        cmap = plt.cm.viridis
#        cmap.set_bad('white')
        plt.imshow(current_magnitude, cmap=cmap, extent=extent, origin='lower')
    else:
        plt.imshow(current_magnitude, cmap='viridis', extent=extent, origin='lower')

    plt.colorbar(label="Модуль тока, А")
    plt.title(f"{title_prefix}\nЧастота: {frequencies[freq_index]:.2f} МГц, Слой: {layer_index}\n ")
    plt.xticks(range(1, Nx + 1))
    plt.yticks(range(1, Ny + 1))
    plt.show()
````

````python
def current_distribution_gif(currents, layer_index, frequencies, Params, title_prefix=""):
    """
    Function to display an animation of current distribution in a given layer of the structure across all frequencies,
    with support for elliptical/spherical masking.

    Parameters:
    currents: array of currents for each frequency.
    layer_index: index of the layer to display (from 0 to Nz-1).
    frequencies: array of frequencies (in MHz).
    Params: dictionary with model parameters, must contain 'shape', 'Orientations', and packing info.
    title_prefix: prefix for the plot title (default is an empty string).
    """
    
    shape_string = Params['Shape']
    shape_list = shape_string.split('x')
    orientations = Params.get('Orientations', 'zyx')
    packing_type = Params['Packing']

    shape_dict = dict(zip(orientations, [int(dim) for dim in shape_list]))
    Nx = shape_dict['x']
    Ny = shape_dict['y']
    Nz = shape_dict['z']

    if layer_index < 0 or layer_index >= Nz:
        print("Индекс слоя вне диапазона.")
        return

    layer_size = Nx * Ny
    extent = [0.5, Nx + 0.5, 0.5, Ny + 0.5]

    # Create mask if packing is Ellipse
    mask = None
    if packing_type == 'Ellipse':
        dz, dy, dx = Params['Dz'], Params['Dy'], Params['Dx']
        Nz_total, Ny_total, Nx_total = Params['N']['z']['nz'], Params['N']['y']['ny'], Params['N']['x']['nx']
        
        R_z, R_y, R_x = (Nz_total-1)*dz/2, (Ny_total-1)*dy/2, (Nx_total-1)*dx/2
        
        # Adjust radii for different packing cases
        if len(orientations) == 1:
            pass
        elif Params['N']['z']['nz'] == Params['N']['y']['nz']:  # center system
            R_z += dz/2
            R_y += dy/2
            R_x += dx/2
        elif Params['N']['z']['nz'] < Params['N']['y']['nz']:  # open system
            R_z += dz 
            R_y += dy
            R_x += dx
        
        # Create mask for cells inside the ellipse
        mask = np.ones((Ny, Nx), dtype=bool)
        for j in range(Ny):
            for i in range(Nx):
                x_pos = i * dx + dx/2
                y_pos = j * dy + dy/2
                z_pos = layer_index * dz + dz/2
                
                distance = ((x_pos-R_x)/R_x)**2 + ((y_pos-R_y)/R_y)**2 + ((z_pos-R_z)/R_z)**2
                if distance > 1.0:
                    mask[j, i] = False

    fig, ax = plt.subplots()
    
    # Set up colormap with white for masked values if needed
    cmap = plt.cm.viridis
    if packing_type == 'Ellipse':
        cmap = cmap.copy()
        cmap.set_bad('white')
    
    im = ax.imshow(np.zeros((Ny, Nx)), cmap=cmap, extent=extent, origin='lower')
    cbar = plt.colorbar(im, ax=ax, label="Модуль тока")
    title = ax.set_title("")
    ax.set_xlabel("X")
    ax.set_ylabel("Y")

    def update(freq_index):
        current_at_freq = currents[freq_index]
        
        start_index = layer_index * layer_size
        end_index = (layer_index + 1) * layer_size
        layer_currents_flat = current_at_freq[start_index:end_index]
        
        layer_currents_matrix = layer_currents_flat.reshape((Ny, Nx))
        
        current_magnitude = np.abs(layer_currents_matrix)
        
        # Apply mask if exists
        if mask is not None:
            current_magnitude = np.where(mask, current_magnitude, np.nan)
        
        im.set_array(current_magnitude)
        title.set_text(f"{title_prefix}Распределение токов\nЧастота: {frequencies[freq_index]:.2f} МГц, Слой: {layer_index}")
        return [im, title]

    ani = FuncAnimation(fig, update, frames=range(len(frequencies)), blit=False)
    plt.close(fig)
    return HTML(ani.to_jshtml())
````

````python
def get_ring_centers(Dx, Dy, Dz, packing, shape_str, orientations):
    """
    Generate an array of coordinates for the centers of all rings in the system.

    Returns
    -------
    coords : np.ndarray
        Array of shape (N, 3), where N is the number of rings.
        Each row is (x, y, z) coordinate of a ring center.
    """

    coords = []
    orientation_list = []
    for o_idx, orientation in enumerate(orientations):
        N = Params['N'][orientation]
        nz, ny, nx = N['nz'], N['ny'], N['nx']
        for k in range(nz):
            for j in range(ny):
                for i in range(nx):
                    x = i * Dx + Dx/2
                    y = j * Dy + Dy/2
                    z = k * Dz + Dz/2
                    coords.append([x, y, z])
                    orientation_list.append(orientation)
    coords = np.array(coords)

    if packing == 'Ellipse':
        Nz = Params['N']['z']['nz']
        Ny = Params['N']['y']['ny']
        Nx = Params['N']['x']['nx']
        Rz, Ry, Rx = (Nz-1)*Dz/2, (Ny-1)*Dy/2, (Nx-1)*Dx/2
        center = np.array([Rx, Ry, Rz]) + np.array([Dx/2, Dy/2, Dz/2])
        mask = (((coords[:,0]-center[0])/Rx)**2 +
                ((coords[:,1]-center[1])/Ry)**2 +
                ((coords[:,2]-center[2])/Rz)**2) <= 1.01
        coords = coords[mask]

    return coords, orientation_list
````

````python
def ring_field_at_point(r0, centers, currents, radii, orientations, Omega, freq_index):
    """
    Calculate the total field at point r0 from all rings.

    Parameters
    ----------
    r0 : array-like, shape (3,)
        Observation point (x, y, z)
    centers : array-like, shape (N, 3)
        Coordinates of ring centers
    currents : array-like, shape (N,)
        Currents in each ring (complex or real)
    radii : float
        Radius of the rings
    orientations : array-like, shape (N,)
        Orientation of each ring ('x', 'y', or 'z')
    Omega : array-like, shape (M,)
        Frequencies (angular frequencies) at which to calculate the field
    freq_index : int
        Index of the frequency in Omega to use for the calculation
    Returns
    -------
    B_total : np.ndarray, shape (3,)
        Total field at r0 (vector)
    """

    omega = Omega[freq_index]
    k = omega / 3e8

    def ring_integrand(phi, center, radius, orientation, I, r0, k):
        if orientation == 'z':
            r_ring = center + radius * np.array([np.cos(phi), np.sin(phi), 0])
            dl = radius * np.array([-np.sin(phi), np.cos(phi), 0])
        elif orientation == 'y':
            r_ring = center + radius * np.array([np.cos(phi), 0, np.sin(phi)])
            dl = radius * np.array([-np.sin(phi), 0, np.cos(phi)])
        elif orientation == 'x':
            r_ring = center + radius * np.array([0, np.cos(phi), np.sin(phi)])
            dl = radius * np.array([0, -np.sin(phi), np.cos(phi)])
        else:
            raise ValueError(f"Unknown orientation: {orientation}")

        R = r0 - r_ring
        R_norm = np.linalg.norm(R)
        prefac = (1j * k * R_norm - 1) * np.exp(1j * k * R_norm) / (R_norm ** 3)
        dB = I * np.cross(dl, R) * prefac
        return dB

    B_total = np.zeros(3, dtype=complex)
    for i in range(len(centers)):
        center = centers[i]
        I = currents[i]
        orientation = orientations[i]

        def integrand(phi):
            return ring_integrand(phi, center, radii, orientation, I, r0, k)

        Bx_real, _ = integrate.quad(lambda phi: np.real(integrand(phi)[0]), 0, 2 * np.pi)
        Bx_imag, _ = integrate.quad(lambda phi: np.imag(integrand(phi)[0]), 0, 2 * np.pi)
        By_real, _ = integrate.quad(lambda phi: np.real(integrand(phi)[1]), 0, 2 * np.pi)
        By_imag, _ = integrate.quad(lambda phi: np.imag(integrand(phi)[1]), 0, 2 * np.pi)
        Bz_real, _ = integrate.quad(lambda phi: np.real(integrand(phi)[2]), 0, 2 * np.pi)
        Bz_imag, _ = integrate.quad(lambda phi: np.imag(integrand(phi)[2]), 0, 2 * np.pi)

        B_total += np.array([
            Bx_real + 1j * Bx_imag,
            By_real + 1j * By_imag,
            Bz_real + 1j * Bz_imag
        ])

    return B_total
````

````python
Params['Dx']=Params['Dz']=Params['Dy']
Params['Orientations'] = 'zyx'
Params['Packing'] = 'Ellipse'
Params['Scattering'] = 'Mie_True'
Params['N'], Params['Shape'] = to3D(6, 6, 6, 'zyx')
data = open_model('DATA', Params)
freq_index = 50
currents = data['Currents'][freq_index]
Omega = data['Omega']
Dx, Dy, Dz = Params['Dx'], Params['Dy'], Params['Dz']
packing = Params['Packing']
shape_str = Params['Shape']
orientations = Params.get('Orientations', 'zyx')
centers, orientations_list = get_ring_centers(Dx, Dy, Dz, packing, shape_str, orientations)

r0 = np.array([0, 0, 0])
radii = Params['Radius']

B = ring_field_at_point(r0, centers, currents, radii, orientations_list, Omega, freq_index)
````

````python
def total_scattering_cross_section_field(B_scat, r_far, Omega, freq_index, H0=1.0, polarization='z'):
    """
    Calculate the total scattering cross-section using the field at a point in the far zone.
    Parameters
    ----------
    ring_field_at_point : function
        Function to calculate the field at a point from the rings.
    r_far : array-like, shape (3,)
        Point in the far zone where the field is calculated.
    centers : array-like, shape (N, 3)
        Coordinates of ring centers.
    currents : array-like, shape (N,)
        Currents in each ring (complex or real).
    radii : float
        Radius of the rings.
    orientations : array-like, shape (N,)
        Orientation of each ring ('x', 'y', or 'z').
    Omega : array-like, shape (M,)
        Frequencies (angular frequencies) at which to calculate the field.
    freq_index : int
        Index of the frequency in Omega to use for the calculation.
    H0 : float, optional
        Reference magnetic field strength (default is 1.0).
    polarization : str, optional
        Polarization of the incident wave ('x', 'y', or 'z', default is 'z').
    """
    omega = Omega[freq_index]
    k = omega / 3e8
    r_abs = np.linalg.norm(r_far)
    if polarization == 'z':
        e0 = np.array([1, 1, 0])
    elif polarization == 'y':
        e0 = np.array([0, 1, 0])
    elif polarization == 'x':
        e0 = np.array([1, 0, 0])
    else:
        raise ValueError("Unknown polarization")
    sigma_tot = (4 * np.pi * r_abs) / (k * H0) * np.imag(np.vdot(e0, B_scat))
    return sigma_tot
````

````python
# Saving polarization with a red dashed line for the variance of frequency
from matplotlib.animation import FuncAnimation

def save_polarization(DATA, name, pol_type = 'Imag'):
    Omega = DATA['Omega']
    Polarization = DATA['Polarization']
    if pol_type == 'Imag':
        Pol = Polarization[:, 0].imag
    else:
        Pol = Polarization[:, 0].real
    fig = plt.figure(figsize=(12, 10))
    plt.plot(Omega/2/np.pi/10**6, Pol, label = f'{pol_type}-polarization', color = 'b')
    omega = plt.axvline(Omega[0]/2/np.pi/1e6, color = 'red', linestyle = '--', label = 'current frequency')
    plt.xlabel('Frequency, MHz')
    plt.ylabel(f'{pol_type}-polarization')
    plt.legend()
    def plot_i(i):
        omega.set_xdata([Omega[i]/2/np.pi/1e6])
        plt.title(f'Frequency = {round(Omega[i]/2/np.pi/1e6, 3)}')
    
    ani = FuncAnimation(fig, plot_i,
                        frames = range(len(Omega)),
                        interval = 50)
    ani.save(f'{name}_{pol_type}.gif', fps=30)
    ani.save(f'{name}_{pol_type}.mp4', fps=30)
    

for shape in Data:
    save_polarization(Data[shape], f'Images/Current_distribution/Form/Zrings/{Params['Packing']}-{shape}-{Params['Solver_type']}', pol_type='Imag')
````

````python
# Calculate polarisation for slices
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.animation as anime
from tqdm import tqdm
import colormaps as cmaps

# Drawing GIFs
current_part = {
    'abs': lambda x: np.abs(x),
    'real': lambda x: np.real(x),
    'imag': lambda x: np.imag(x),
    'angle': lambda x: np.angle(x)
}

def SaveDistribution(Data, rtype, slice_deep, slice_or, name, folder = '', mode = 'abs'):
    Currents_4D = Data['Currents']
    Omega = np.array(Data["Omega"])
    part = current_part[mode]
    Params = Data['Params']
    N = Params['N']

    start = 0
    for ring_type in N:
        if ring_type != rtype:
            start += np.prod([N[ring_type][j] for j in N[ring_type]])
        else:
            break
    end = start + np.prod([N[rtype][j] for j in N[rtype]])

    rslice = [slice(slice_deep * (slice_or =='z'), (slice_deep+1)*(slice_or=='z') + N[rtype]['nz'] * (slice_or != 'z'), 1),
      slice(slice_deep * (slice_or =='y'), (slice_deep+1)*(slice_or=='y') + N[rtype]['ny'] * (slice_or != 'y'), 1),
      slice(slice_deep * (slice_or =='x'), (slice_deep+1)*(slice_or=='x') + N[rtype]['nx'] * (slice_or != 'x'), 1)]
    
    shape = [len(Omega)]+[N[rtype][j] for j in N[rtype]]
    squeeze_or = 1 if slice_or == 'z' else 2 if slice_or == 'y' else 3

    # Find currents on slice

    Currents2D = Currents_4D[:,start:end].reshape(shape)[:, *rslice]*Params['P_0z']
    
    # convert 3D data to 2D along slice_or axis
    Currents2D = np.squeeze(part(Currents2D), axis = squeeze_or) 

    # Contourf can't plot 1D data, so we need to repeat it
    if Currents2D.shape[1] == 1:
        Currents2D = np.repeat(Currents2D, 2, axis = 1)
    if Currents2D.shape[2] == 1:
        Currents2D = np.repeat(Currents2D, 2, axis = 2)
    
    
    # Calculate extremums for each slice
    MaxCurrents = Currents2D.max()
    MinCurrents = Currents2D.min()
    MaxDeltaCurrents = MaxCurrents - MinCurrents

    fig_currents, ax_currents = plt.subplots(figsize = (12, 10))
    
    if slice_or == 'z':
        n_1, n_2 = r'$n_x$', r'$n_y$'
    elif slice_or == 'y':
        n_1, n_2 = r'$n_x$', r'$n_z$'
    else:
        n_1, n_2 = r'$n_y$', r'$n_z$'
    
    X, Y = list(range(N[rtype][n_1[1]+n_1[3]])), list(range(N[rtype][n_2[1]+n_2[3]]))
    if len(X) == 1:
        X = [0, 1]
    if len(Y) == 1:
        Y = [0, 1]

    def plot_GIF(i):
        plt.clf()
        
        fig_currents.set_size_inches((12, 10))

        Currents = Currents2D[i]
        omega = Omega[i]
        
        # Decommend for level lines on the plot

        Lines = plt.contour(X, Y, Currents, np.min([len(X), len(Y), 6]), colors = 'black', alpha = 0.01)
        
        Imax, Imin = Currents.max(), Currents.min()
        DeltaCurrents = Imax - Imin
        
        amplitude = DeltaCurrents/MaxDeltaCurrents/2
        if amplitude == 0:
            amplitude = 0.01
        if MaxDeltaCurrents - DeltaCurrents == 0:
            middlepos = 0.5
        else:
            middlepos = (Imin-MinCurrents)/(MaxDeltaCurrents - DeltaCurrents)
        ticksnumber = round(amplitude * 10) + 1
        
        ticks = list(map(lambda x: round(np.real(x), 2), np.linspace(Imax, Imin, ticksnumber)))

        plt.xlabel(n_1, fontsize = 24)
        plt.ylabel(n_2, fontsize = 24)
        
        # Choosing color map

        #plt.contourf(X, Y, Currents, 100, cmap=cmaps.voltage)
        #plt.pcolormesh(X, Y, Currents.T, cmap=cmaps.voltage, shading='auto')
        plt.imshow(Currents, cmap=cmaps.voltage, origin='lower', extent = [X[0], X[-1]+1, Y[0], Y[-1]+1])
        
        plt.clabel(Lines, inline=1, fontsize=15)
        cbar = plt.colorbar(label = r'$\chi$', fraction = 0.03, pad = 0.05, 
                          shrink = 10, aspect = amplitude/0.03, anchor = (0, middlepos),
                          ticks = ticks)
        cbar.set_label(r'$\chi$', fontsize = 24)
        cbar.ax.tick_params(labelsize=20)

        plt.title(f"{slice_or} slice on {slice_deep} layer for {rtype}-ring: $f =${round(omega/2/np.pi/10 ** 6, 3)}, MHz",
                  fontsize = 24,
                  pad = 40)
        
        plt.xticks(fontsize = 20)
        plt.yticks(fontsize = 20)
        plt.axis('equal')
        for x in X:
            plt.axvline(x, color = 'black')
        #plt.axvline(X[-1]+1, color = 'black')
        for y in Y:
            plt.axhline(y, color = 'black')
        #plt.axhline(Y[-1]+1, color = 'black')
    
    ani = anime.FuncAnimation(fig_currents, plot_GIF,
                              frames = tqdm(list(range(len(Omega)))),
                              interval = 50
                              )
    if not os.path.exists(folder):
        os.makedirs(folder)
    ani.save(f'{folder}/{name}.gif', fps=30)
    ani.save(f'{folder}/{name}.mp4', fps=30)

folder = 'Images/Current_distribution/Form/Zrings'
for shape in Data:
    data = Data[shape]
    l = int(shape.split('_')[1].split('x')[0])
    for rtype in ['z']:
        for slice_or in ['z', 'y']:
                slice_deep = 24
                print('Saving: ', shape, rtype, slice_or, slice_deep)
                for mode in ['abs', 'angle']:
                    SaveDistribution(data, rtype, slice_deep, slice_or, f'Plots/Current_distribution/Form/{Params['Packing']}-{shape}-{rtype}ring-{slice_or}slice-{slice_deep}-{Params['Solver_type']}-{mode}', mode = mode)
````

````python
# Calculate and compare polarisation for different structures

import matplotlib.pyplot as plt
import numpy as np
import matplotlib.animation as anime
from tqdm import tqdm

# Drawing GIFs

''' 
- Datas_sets is a dictionary with keys as names of datasets and values as datasets
- rtypes is a list of ring types
- slice_ors is a list of slice orientations
- slice_deeps is a dictionary with keys as tuples (name, rtype, slice_or) and values as lists of slice deeps,
number of slice deeps should be the same for each id
- name is a name of the GIF 
'''
def SaveDistribution(Data_sets, rtypes, slice_ors, slice_deeps, Name):
    
    # Get Data for plotting
    Currents2D = {}
    Omega = {}
    MaxCurrents = {}
    MinCurrents = {}
    MaxDeltaCurrents = {}
    n_1, n_2 = {}, {}
    X, Y = {}, {}

    grid_num = 0
    for name in Data_sets:
        Data = Data_sets[name]
        Omega[name] = np.array(Data["Omega"])
        Currents_4D = np.array(Data["RealCurrents"], dtype = complex) + 1j * np.array(Data["ImagCurrents"], dtype= complex)

        Params = Data['Params']

        Dz, Dy, Dx = Params['Dz'], Params['Dy'], Params['Dx']
        Radius = Params['Radius']
        N = Params['N']
        H_0z = 1

        for rtype in rtypes:
            start = 0
            # Calculate start and end positions for each ring type
            for ring_type in N:
                if ring_type != rtype:
                    start += np.prod([N[ring_type][j] for j in N[ring_type]])
                else:
                    break
            end = start + np.prod([N[rtype][j] for j in N[rtype]])
            
            for slice_or in slice_ors:
                slice_id = (name, rtype, slice_or)
                for slice_deep in slice_deeps[slice_id]:
                    # Calculate position in plotting grid
                    data_pos = 1 + list(Data_sets.keys()).index(name)
                    slice_deep_pos = list(slice_deeps[slice_id]).index(slice_deep) * len(Data_sets)
                    slice_pos = list(slice_ors).index(slice_or) * len(Data_sets) * len(slice_deeps[slice_id])
                    rtype_pos = list(rtypes).index(rtype) * len(Data_sets) * len(slice_ors) * len(slice_deeps[slice_id])

                    grid_num = data_pos + slice_pos + slice_deep_pos + rtype_pos
                    Id = (name, rtype, slice_or, slice_deep, grid_num)
                    
                    # Get current slice for choosen parameters
                    rslice = [slice(slice_deep * (slice_or =='z'), (slice_deep+1)*(slice_or=='z') + N[rtype]['nz'] * (slice_or != 'z'), 1),
                      slice(slice_deep * (slice_or =='y'), (slice_deep+1)*(slice_or=='y') + N[rtype]['ny'] * (slice_or != 'y'), 1),
                      slice(slice_deep * (slice_or =='x'), (slice_deep+1)*(slice_or=='x') + N[rtype]['nx'] * (slice_or != 'x'), 1)]

                    Currents2D[Id] = np.real(np.squeeze(Currents_4D[:,start:end].reshape([len(Omega[name])]+ [N[rtype][j] for j in N[rtype]])[:, *rslice]))*np.pi * Radius ** 2 /Dz/Dy/Dx/H_0z

                    # Calculate extremums for each slice
                    MaxCurrents[Id] = Currents2D[Id].max()
                    MinCurrents[Id] = Currents2D[Id].min()
                    MaxDeltaCurrents[Id] = MaxCurrents[Id] - MinCurrents[Id]

                    n_1[Id], n_2[Id] = tuple([j for j in N[rtype].keys() if j != f'n{slice_or}'])

                    X[Id], Y[Id] = list(range(N[rtype][n_1[Id]])), list(range(N[rtype][n_2[Id]]))
        

    fig, ax = plt.subplots()

    # Numbers of plots in x and y axes

    xn = len(Data_sets) * len(slice_deeps[slice_id])
    yn = len(rtypes) * len(slice_ors) 

    # Figsize
    xsize = xn*12
    ysize = yn*10

    def plot_GIF(i):
        # Clearing figure and set figsize
        plt.clf()
        fig.set_size_inches(xsize, ysize)
        for Id in Currents2D:
            name, rtype, slice_or, slice_deep, grid_num = Id
            
            plt.subplot(yn, xn, grid_num)
            Currents = Currents2D[Id][i]    
            omega = Omega[name][i]

            # Set axes scale and avoid extra long plots

            # Draw same-value lines
            Lines = plt.contour(X[Id], Y[Id], Currents.T, 6, colors = 'black')

            # Calculate max currents to normilze it for spectral max
            Imax, Imin = Currents.max(), Currents.min()
            DeltaCurrents = Imax - Imin
            
            # Calculate length of colorbar
            amplitude = DeltaCurrents/MaxDeltaCurrents[Id]

            if abs((Imin - MinCurrents[Id])/Imin) < 1e-3 and abs((Imax - MaxCurrents[Id])/MaxCurrents[Id]) < 1e-3:
                middlepos = 0
            else:
                middlepos = (Imin-MinCurrents[Id])/(MaxDeltaCurrents[Id] - DeltaCurrents)

            # Calculate number of ticks depended on coloubar length
            ticksnumber = round(amplitude * 10) + 1
            ticks = list(map(lambda x: round(np.real(x), 2), np.linspace(Imax, Imin, ticksnumber)))

            plt.xlabel(n_1[Id], fontsize = 15)
            plt.ylabel(n_2[Id], fontsize = 15)

            # Draw colormap
            plt.contourf(X[Id], Y[Id], Currents.T, 100, cmap=plt.cm.hot)
            plt.clabel(Lines, inline=1, fontsize=15)
            plt.colorbar(label = r'$\chi$', fraction = 0.03, pad = 0.05, 
                              shrink = 10, aspect = amplitude/0.03, anchor = (0, middlepos),
                              ticks = ticks)
            scale = len(Y[Id])/len(X[Id])

            if scale < 1/3:
                ax.set_aspect(1/3)
            elif scale > 3:
                ax.set_aspect(3)
            else:
                ax.set_aspect(scale)
            
            plt.title(f"{name} {slice_or} slice on {slice_deep} layer for {rtype}-ring: $f =${round(omega/2/np.pi/10 ** 6, 3)}, MHz", fontsize = 15)

    # Make and save animation object
    ani = anime.FuncAnimation(fig, plot_GIF,
                              frames = list(range(len(Omega[name]))),
                              interval = 50
                              )

    ani.save(f'{Name}.gif', writer='pillow', fps=24, dpi=80)

Data_sets = {'Needle': Data['10x10x10']}#, 'Straight': Data_straight}
rtypes = ('z')
slice_ors = ('y')
slice_deeps = {}
for name in Data_sets:
    for rtype in rtypes:
        for slice_or in slice_ors:
            slice_deeps[(name, rtype, slice_or)] = [0]
SaveDistribution(Data_sets, rtypes, slice_ors, slice_deeps, f'Plots/Current_distribution/Form/Errors')        
````
